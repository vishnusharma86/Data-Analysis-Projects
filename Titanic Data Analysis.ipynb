```python
g=df["gender"] # count male and female passengers using for loop
male=0
female=0
for i in g:
    if i=="male":
        male=male+1
    if i=="female":
        female=female+1
print("MALE :",male)
print("FEMALE",female)
```

```python
clas=df["Pclass"] # count passenger by class
class1=0
class2=0
class3=0
for i in clas:
    if i==1:
        class1=class1+1
    elif i==2:
        class2=class2+1
    else:
        class3=class3+1
print(class1)
print(class2)
print(class3)
```

```python
t=df["Fare"] # check cheapest and expensive tickets price's
cheap=min(t)
expensive=max(t)
print(cheap)
print(expensive)
```

```python
entry=df["Embarked"] # passengers count by board
S=0
C=0
Q=0
for i in entry:
    if i=="S":
        S=S+1
    elif i=="C":
        C=C+1
    elif i=="Q":
        Q=Q+1
print("S",S)
print("C",C)
print("Q",Q)
```
