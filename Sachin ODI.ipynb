```python
import pandas as pd
df=pd.read_csv("sachin_odi.csv")
df.head(10)
```

```python
df["batting_score"] = pd.to_numeric(df["batting_score"], errors='coerce')
```

```python
# Q1. Jab India jeet ti thi aur jab harti thi — dono mein Sachin ka average batting score compare karo.

df.groupby("match_result").batting_score.agg(["mean"])
```

```python
# Q2. Har team (opposition) ke against kitne matches khele gaye? Sabse zyada matches kis team ke against the?

df.groupby("opposition").opposition.agg(["count"]) 
```

```python
# Q3. Toss jeeta ya haara — dono cases mein matches ka count karo. Toss ka match result pe kya asar tha?

df.groupby(["toss"]).toss.agg(["count"]) 
```

```python
df.groupby("match_result").match_result.agg(["count"]) #Que3
```

```python
# Q4. Pehli innings mein batting ki ya doosri mein — dono mein average score compare karo.

df.groupby("batting_innings").batting_score.agg(["mean","min","max"])
```

```python
# Q5. Har team ke against Sachin ka average batting score kya tha? Kis team ke against sabse zyada runs banaye?

df.groupby("opposition").batting_score.agg(["mean"])
```

```python
# Q6. Match result aur toss dono ka combination groupby karo — 4 groups ka match count compare karo.

df.groupby(["match_result","toss"]).match_result.agg(["count"]).head(16)
```

```python
# Q7. Har team ke against jeetne wale aur haarne wale matches alag count karo.

df.groupby(["opposition","match_result"]).match_result.agg(["count"])
```

```python
# Q8. Batting innings (1st/2nd) aur match result ka combination — 2nd innings mein batting karne pe zyada jeete?

df.groupby(["batting_innings","match_result"]).match_result.agg(["count"]) 
```
