```python
import pandas as pd
df=pd.read_csv("salaray_projects.csv")
df
```

```python
df.groupby("Projects").Salary.mean() # Average Salary By Projects
```

```python
max_sal=df["Salary"].max() # Max Salary And Minimum Salary
min_sal=df["Salary"].min()
print("Maximun Salary :",max_sal)
print()
print("Minimun Salary :",min_sal)
```

```python
import numpy as np
import statistics as st 
a=df["WorkExp"]
print(st.quantiles(a))
```

```python
three=df[df["WorkExp"]<3] # Work experince 0-3
seven=df[(df["WorkExp"]>=3) & (df["WorkExp"]<7)] # Work experince 3-7
ten=df[df["WorkExp"]>=7] # Work experince 7+

print("Work Exp 0-3 Average Salary :",three.Salary.mean())
print()
print("Work Exp 3-7 Average Salary :",seven.Salary.mean())
print()
print("Work Exp 7+ Average Salary :",ten.Salary.mean())
```

```python
under50=df[df["Salary"]<=50000]
above50=df[df["Salary"]>50000]
print("Average Work Exp Of Above 50 Thousand Salaried Employee :",above50.WorkExp.mean())
print("Average Work Exp Of Under 50 Thousand Salaried Employee :",under50.WorkExp.mean())
```
