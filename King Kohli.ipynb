```python
df["Runs"] = df["Runs"].str.replace("*", "", regex=False) #Que2
df["Runs"] = pd.to_numeric(df["Runs"], errors='coerce')
df.groupby("Ground")["Runs"].agg(["count", "mean"]).round(2)
```
