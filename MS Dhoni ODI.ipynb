```python
import pandas as pd
import numpy as np
```

```python
# total number of run in odi  
# total average of run in odi 
# total not out in odi  
# total out in odi      
# total hundreds in odi  
# total fifty in odi     
# highest run in odi     
# total out on duck      
```

```python
#1. total number of run in odi
d=df.Runs
d=d.replace("DNB",0,regex=True)
d=d.replace("[^0-9]","",regex=True)
s=np.array(d,dtype=np.int32)
total_run=s.sum()
print("Total Runs In ODI :",total_run)
```

```python
#2. total average of run in odi 
df=pd.read_csv(r"C:\Users\eichi\OneDrive\Desktop\ABC\MSDhoni_ODI_Stats.csv")
d=df.Runs
dnb=0
no=0
for i in d:
    if "DNB" in i:
        dnb=dnb+1
notout=d.replace("[0-9]","",regex=True)
for j in notout:
    if j=="*":
        no=no+1
print("DNB :",dnb)
print("NotOut :",no)
print("Average Run :",total_run/(len(d)-dnb-no))
```

```python
#3. total not out in odi  
#4. total out in odi  
d=df.Runs
d=d.replace("DNB",0,regex=True)
d=d.replace("[0-9]","",regex=True)
notout=0
out=0
for i in d:
    if i=="*":
        notout=notout+1
    elif i=="":
        out=out+1
print("Total NotOut's :",notout)
print("Total Out's :",out)
```

```python
#5. total hundreds in odi  
#6. total fifty in odi     
#7. highest run in odi     
#8. total out on duck
run=df.Runs
run=run.replace("DNB",0,regex=True)
run=run.replace("[^0-9]","",regex=True)
run=pd.to_numeric(run)
half=0
century=0
duck=0
for i in run:
    if i>=50 and i<100:
        half=half+1
    elif i>=100:
        century=century+1
for j in run:
    if j==0:
        duck=duck+1
print("Half Centuries :",half)
print("Centuries :",century)
print("Highest Run :",run.max())
print("Total Out On Duck :",duck)
```
