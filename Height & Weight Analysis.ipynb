```python
import pandas as pd
df=pd.read_csv("weight-height.csv")
df
```

```python
df.groupby("Gender").mean() #Average Height And Weight of Gender
```

```python
df.groupby("Gender").Weight.agg(["min","max"]) # min and max weight of gender
```

```python
short=df[df["Height"]<160]  # splits in three categories for counting
average=df[(df["Height"]>=160) & (df["Height"]<180)]
tall=df[df["Height"]>=180]

print("Short :",short.Gender.count())
print()
print("Average :",average.Gender.count())
print()
print("Tall :",tall.Gender.count())
```

```python
print("Short :",short.Weight.mean())  # Average Weights of categories hight
print()
print("Average :",average.Weight.mean())
print()
print("Tall :",tall.Weight.mean())
```

```python
df.groupby("Gender").Weight.std() #Standard Deviation
```
