```python
## Groupby And Aggregation
```

```python
df.groupby("region").purchase_amount.agg(["mean"]) # avg purchase amounts in each regions
```

```python
df.groupby("region").user_id.agg(["count"]) # Count the number of customers in each region.
```

```python
df.groupby("region").purchase_frequency.agg(["max","mean"]) # the highest average purchase frequency region
```

```python
df.groupby("region").annual_income.agg(["min","max"])  # the maximum and minimum annual income for each region
```

```python
score=df["loyalty_score"] # Divide customers into three loyalty categories and count customers in each category
low=0 # 0-4
medium=0 # 4-7
high=0 # 7-10
for i in score:
    if i>=0 and i<4:
        low=low+1
    elif i>=4 and i<7:
        medium=medium+1
    elif i>=7 and i<=10:
        high=high+1
print("Low Loyalty Score customers :",low)
print("Medium Loyalty Score customers :",medium)
print("High Loyalty Score customers :",high)
```

```python
df.groupby("age").purchase_amount.agg(["sum"]) # the total purchase amount for each age group.
```

```python
df.groupby("region").loyalty_score.agg(["mean"]) # region has the highest average loyalty score - West
```

```python
income=df["annual_income"]
lowincome=[]
highincome=[]
for i in income:
    if i<50000:
        lowincome.append(i)
    elif i>=50000:
        highincome.append(i)
print(np.mean(lowincome))
print(np.mean(highincome))
```
