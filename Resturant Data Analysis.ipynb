```python
import pandas as pd
```

```python
df=pd.read_csv("restaurant.csv")
```

```python
avg_male_tip=df[df.gender=="Male"].tip.mean() # Average Male Tip
avg_female_tip=df[df.gender=="Female"].tip.mean() # Average Female Tip
print(avg_male_tip)
print(avg_female_tip)
```

```python
df.groupby(by=["day"]).total_bill.agg(["mean"]) # Per Day Average Bill
```

```python
df.groupby(by=["smoker"]).tip.agg(["mean"]) # Tip By Smoker and Non Smoker
```

```python
a=df["time"] # Count Dinner And Lunch
dinner=0 
lunch=0
for i in a:
    if i=="Dinner":
        dinner=dinner+1
    elif i=="Lunch":
        lunch=lunch+1
print(dinner)
print(lunch)
```

```python
df.groupby('day')[['total_bill', 'size']].agg({'total_bill': 'count', 'size': 'mean'}) # per day customer count and avg table size
```

```python
df.groupby(by=["gender","smoker"]).tip.agg(["mean"]) # Gender And Smoker Wise Average Tip
```

```python
df.groupby(by="size").total_bill.agg(["mean"]) # Average Total Bill By TableSize
```

```python
df.groupby("day").tip.agg(["sum"])
```
